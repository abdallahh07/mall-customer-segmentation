## Overview

This is a hands-on practice project — my first time building a clustering model. Using the classic Mall Customers dataset (200 customers, membership card data), the goal is to segment customers into distinct groups a marketing team could target differently.

**What this notebook covers:**
1. Load and explore the data
2. Select and scale relevant features (Annual Income, Spending Score)
3. Use the elbow method to choose the number of clusters
4. Fit K-Means and visualize the resulting segments
5. Interpret what each customer segment represents
